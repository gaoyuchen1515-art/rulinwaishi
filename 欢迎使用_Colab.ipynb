```
import torch
print(f"PyTorch版本: {torch.__version__}")
print(f"是否有GPU: {torch.cuda.is_available()}")
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"使用设备: {device}")

!pip install scikit-learn matplotlib seaborn torchsummary -q

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import time
import random
from tqdm import tqdm
from torchsummary import summary

import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, random_split
from torchvision import datasets, transforms
from torchvision.models import swin_t, Swin_T_Weights

from sklearn.metrics import accuracy_score, f1_score
from sklearn.manifold import TSNE
from sklearn.utils import resample

seed = 42
torch.manual_seed(seed)
torch.cuda.manual_seed(seed)
np.random.seed(seed)
random.seed(seed)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 12
sns.set_style("whitegrid")

raw_train_dataset = datasets.CIFAR10(
    root='./data', train=True, download=True, transform=transforms.ToTensor()
)

raw_test_dataset = datasets.CIFAR10(
    root='./data', train=False, download=True, transform=transforms.ToTensor()
)
CIFAR10_classes = ['airplane', 'automobile', 'bird', 'cat', 'deer',
                   'dog', 'frog', 'horse', 'ship', 'truck']

train_data = torch.stack([img for img, _ in raw_train_dataset])

mean = train_data.mean(dim=(0, 2, 3)).numpy()
std = train_data.std(dim=(0, 2, 3)).numpy()
print(f"训练集RGB通道均值: {mean.round(4)}")
print(f"训练集RGB通道标准差: {std.round(4)}")

train_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=10),
    transforms.Normalize(mean=mean, std=std)
])

val_test_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(mean=mean, std=std)
])

swin_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Resize((224, 224)),
    transforms.Normalize(mean=mean, std=std)
])

full_train_dataset = datasets.CIFAR10(
    root='./data', train=True, download=False, transform=train_transform
)

train_size = 45000
val_size = 5000
train_dataset, val_dataset = random_split(full_train_dataset, [train_size, val_size])

test_dataset = datasets.CIFAR10(
    root='./data', train=False, download=False, transform=val_test_transform
)

swin_train_dataset = datasets.CIFAR10(root='./data', train=True, download=False, transform=swin_transform)
swin_train_dataset, swin_val_dataset = random_split(swin_train_dataset, [train_size, val_size])
swin_test_dataset = datasets.CIFAR10(root='./data', train=False, download=False, transform=swin_transform)

BATCH_SIZE = 16

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=2)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2)

swin_train_loader = DataLoader(swin_train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=2)
swin_val_loader = DataLoader(swin_val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2)
swin_test_loader = DataLoader(swin_test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2)


print(f"\n数据加载完成：")
print(f"CNN训练集: {len(train_dataset)} 样本 | 验证集: {len(val_dataset)} 样本 | 测试集: {len(test_dataset)} 样本")
print(f"Swin-T训练集: {len(swin_train_dataset)} 样本 | 验证集: {len(swin_val_dataset)} 样本 | 测试集: {len(swin_test_dataset)} 样本")

vgg11_config = [
    (1, 64, 3, 1, 1), 'M',
    (1, 128, 3, 1, 1), 'M',
    (2, 256, 3, 1, 1), 'M',
    (2, 512, 3, 1, 1), 'M',
    (2, 512, 3, 1, 1), 'M'
]

class VGG11(nn.Module):
    def __init__(self, config, num_classes=10):
        super(VGG11, self).__init__()
        self.features = self._build_features(config)
        self.classifier = nn.Sequential(
            nn.Linear(512 * 1 * 1, 512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.5),
            nn.Linear(512, 256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.5),
            nn.Linear(256, num_classes)
        )

        self._initialize_weights()

    def _build_features(self, config):
        layers = []
        in_channels = 3
        for cfg in config:
            if cfg == 'M':
                layers.append(nn.MaxPool2d(kernel_size=2, stride=2))
            else:
                conv_layers, out_channels, k, s, p = cfg
                for _ in range(conv_layers):
                    layers.append(nn.Conv2d(in_channels, out_channels, kernel_size=k, stride=s, padding=p))
                    layers.append(nn.BatchNorm2d(out_channels))  # BN加速训练
                    layers.append(nn.ReLU(inplace=True))
                    in_channels = out_channels
        return nn.Sequential(*layers)

    def forward(self, x):
        x = self.features(x)
        x = x.view(x.size(0), -1)
        x = self.classifier(x)
        return x

    def extract_penultimate_feature(self, x):
        x = self.features(x)
        x = x.view(x.size(0), -1)

        for layer in list(self.classifier)[:-1]:
            x = layer(x)
        return x

    def _initialize_weights(self):
        for m in self.modules():
            if isinstance(m, nn.Conv2d):
                nn.init.kaiming_normal_(m.weight, mode='fan_out', nonlinearity='relu')
                if m.bias is not None:
                    nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.BatchNorm2d):
                nn.init.constant_(m.weight, 1)
                nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.Linear):
                nn.init.normal_(m.weight, 0, 0.01)
                nn.init.constant_(m.bias, 0)

vgg_model = VGG11(vgg11_config, num_classes=10).to(device)

summary(vgg_model, input_size=(3, 32, 32))

swin_model = swin_t(weights=Swin_T_Weights.IMAGENET1K_V1).to(device)

in_features = swin_model.head.in_features
swin_model.head = nn.Linear(in_features=in_features, out_features=10).to(device)

def swin_extract_penultimate_feature(model, x):
    x = model.features(x)
    x = model.norm(x)
    x = x.mean(dim=[2, 3])
    return x

print("Swin-T分类头替换完成：")
print(swin_model.head)

summary(swin_model, input_size=(3, 224, 224))

LEARNING_RATE = 1e-4
OPTIMIZER = "Adam"
EPOCHS = 20
LOSS_FUNC = nn.CrossEntropyLoss()

print(f"超参配置（已记录）：")
print(f"Batch Size: {BATCH_SIZE} | Learning Rate: {LEARNING_RATE}")
print(f"Optimizer: {OPTIMIZER} | Epochs: {EPOCHS} | Loss Function: CrossEntropyLoss")


def train_model(model, train_loader, val_loader, epochs, lr, optimizer_name, loss_func, device):

    if optimizer_name == "Adam":
        optimizer = optim.Adam(model.parameters(), lr=lr)
    elif optimizer_name == "SGD":
        optimizer = optim.SGD(model.parameters(), lr=lr, momentum=0.9)
    else:
        raise ValueError("优化器仅支持Adam/SGD")

    train_losses = []
    val_losses = []

    epoch_times = []

    model.train()
    for epoch in range(epochs):
        start_time = time.time()
        train_loss = 0.0

        for imgs, labels in tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs}"):
            imgs, labels = imgs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(imgs)
            loss = loss_func(outputs, labels)
            loss.backward()
            optimizer.step()
            train_loss += loss.item() * imgs.size(0)

        train_loss /= len(train_loader.dataset)
        train_losses.append(train_loss)

        model.eval()
        val_loss = 0.0
        with torch.no_grad():
            for imgs, labels in val_loader:
                imgs, labels = imgs.to(device), labels.to(device)
                outputs = model(imgs)
                loss = loss_func(outputs, labels)
                val_loss += loss.item() * imgs.size(0)
        val_loss /= len(val_loader.dataset)
        val_losses.append(val_loss)


        epoch_time = time.time() - start_time
        epoch_times.append(epoch_time)


        print(f"Epoch {epoch+1} | 训练损失: {train_loss:.4f} | 验证损失: {val_loss:.4f} | 耗时: {epoch_time:.2f}s")
        model.train()

    avg_epoch_time = np.mean(epoch_times)
    print(f"\n训练完成！平均每轮训练时间: {avg_epoch_time:.2f}s")
    return model, train_losses, val_losses, avg_epoch_time

def evaluate_model(model, test_loader, device, is_swin=False):
    model.eval()
    all_preds = []
    all_labels = []
    with torch.no_grad():
        for imgs, labels in tqdm(test_loader, desc="评估测试集"):
            imgs, labels = imgs.to(device), labels.to(device)
            outputs = model(imgs)
            preds = torch.argmax(outputs, dim=1)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    accuracy = accuracy_score(all_labels, all_preds) * 100
    f1 = f1_score(all_labels, all_preds, average='macro') * 100
    print(f"测试集准确率: {accuracy:.2f}% | 宏平均F1: {f1:.2f}%")
    return accuracy, f1, all_labels, all_preds

def extract_features(model, test_loader, device, is_swin=False, sample_size=1000):
    model.eval()
    all_features = []
    all_labels = []
    with torch.no_grad():
        for imgs, labels in tqdm(test_loader, desc="提取特征"):
            imgs, labels = imgs.to(device), labels.to(device)
            if is_swin:

                features = swin_extract_penultimate_feature(model, imgs)
            else:

                features = model.extract_penultimate_feature(imgs)
            all_features.extend(features.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    features_np = np.array(all_features)
    labels_np = np.array(all_labels)
    features_sampled, labels_sampled = resample(features_np, labels_np, n_samples=sample_size, random_state=seed)
    print(f"特征提取完成！采样{sample_size}个样本，特征维度: {features_sampled.shape[1]}")
    return features_sampled, labels_sampled

def tsne_visualization(features, labels, title, save_path):

    tsne = TSNE(n_components=2, random_state=seed, perplexity=30, n_iter=1000)
    features_2d = tsne.fit_transform(features)

    plt.figure(figsize=(10, 8))
    for c in range(10):
        mask = labels == c
        plt.scatter(features_2d[mask, 0], features_2d[mask, 1], label=CIFAR10_classes[c], s=10, alpha=0.8)
    plt.title(title, fontsize=14)
    plt.xlabel("t-SNE Dimension 1")
    plt.ylabel("t-SNE Dimension 2")
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.savefig(save_path, dpi=300)
    plt.show()
    return features_2d

def plot_loss_curve(train_losses, val_losses, title, save_path):
    plt.figure(figsize=(10, 6))
    plt.plot(range(1, EPOCHS+1), train_losses, label="训练损失", linewidth=2)
    plt.plot(range(1, EPOCHS+1), val_losses, label="验证损失", linewidth=2)
    plt.title(title, fontsize=14)
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.savefig(save_path, dpi=300)
    plt.show()

print("="*50 + " 开始训练VGG11 " + "="*50)
vgg_trained, vgg_train_losses, vgg_val_losses, vgg_avg_epoch_time = train_model(
    model=vgg_model,
    train_loader=train_loader,
    val_loader=val_loader,
    epochs=EPOCHS,
    lr=LEARNING_RATE,
    optimizer_name=OPTIMIZER,
    loss_func=LOSS_FUNC,
    device=device
)

vgg_accuracy, vgg_f1, vgg_test_labels, vgg_test_preds = evaluate_model(
    model=vgg_trained,
    test_loader=test_loader,
    device=device,
    is_swin=False
)

plot_loss_curve(
    train_losses=vgg_train_losses,
    val_losses=vgg_val_losses,
    title="VGG11 训练损失 vs 验证损失",
    save_path="vgg_loss_curve.png"
)

vgg_features, vgg_tsne_labels = extract_features(
    model=vgg_trained,
    test_loader=test_loader,
    device=device,
    is_swin=False,
    sample_size=1000
)
tsne_visualization(
    features=vgg_features,
    labels=vgg_tsne_labels,
    title="VGG11 倒数第二层特征 t-SNE可视化（CIFAR-10，1000样本）",
    save_path="vgg_tsne.png"
)

print("="*50 + " 开始训练Swin-T " + "="*50)
swin_trained, swin_train_losses, swin_val_losses, swin_avg_epoch_time = train_model(
    model=swin_model,
    train_loader=swin_train_loader,
    val_loader=swin_val_loader,
    epochs=EPOCHS,
    lr=LEARNING_RATE,
    optimizer_name=OPTIMIZER,
    loss_func=LOSS_FUNC,
    device=device
)

swin_accuracy, swin_f1, swin_test_labels, swin_test_preds = evaluate_model(
    model=swin_trained,
    test_loader=swin_test_loader,
    device=device,
    is_swin=True
)

plot_loss_curve(
    train_losses=swin_train_losses,
    val_losses=swin_val_losses,
    title="Swin-T 训练损失 vs 验证损失",
    save_path="swin_loss_curve.png"
)

swin_features, swin_tsne_labels = extract_features(
    model=swin_trained,
    test_loader=swin_test_loader,
    device=device,
    is_swin=True,
    sample_size=1000
)
tsne_visualization(
    features=swin_features,
    labels=swin_tsne_labels,
    title="Swin-T 倒数第二层特征 t-SNE可视化（CIFAR-10，1000样本）",
    save_path="swin_tsne.png"
)

def count_parameters(model):

    return sum(p.numel() for p in model.parameters() if p.requires_grad) / 1e6

vgg_params = count_parameters(vgg_trained)
swin_params = count_parameters(swin_trained)

print("="*80)
print("CNN(VGG11) vs Swin-T 最终性能+效率对比（报告核心数据）")
print("="*80)
print(f"1. 性能指标（测试集）：")
print(f"   VGG11: 准确率 = {vgg_accuracy:.2f}% | 宏平均F1 = {vgg_f1:.2f}%")
print(f"   Swin-T: 准确率 = {swin_accuracy:.2f}% | 宏平均F1 = {swin_f1:.2f}%")
print(f"2. 效率指标：")
print(f"   VGG11: 可训练参数量 = {vgg_params:.2f}M | 平均每轮训练时间 = {vgg_avg_epoch_time:.2f}s")
print(f"   Swin-T: 可训练参数量 = {swin_params:.2f}M | 平均每轮训练时间 = {swin_avg_epoch_time:.2f}s")
print("="*80)

comparison_data = {
    "model": ["VGG11", "Swin-T"],
    "accuracy(%)": [vgg_accuracy, swin_accuracy],
    "f1_macro(%)": [vgg_f1, swin_f1],
    "params(M)": [vgg_params, swin_params],
    "avg_epoch_time(s)": [vgg_avg_epoch_time, swin_avg_epoch_time]
}
import pandas as pd
df = pd.DataFrame(comparison_data)
df.to_csv("model_comparison.csv", index=False)
print("对比数据已保存为model_comparison.csv，可直接导入报告！")

from google.colab import files
files.download("vgg_loss_curve.png")
files.download("swin_loss_curve.png")
files.download("vgg_tsne.png")
files.download("swin_tsne.png")
files.download("model_comparison.csv")
```
